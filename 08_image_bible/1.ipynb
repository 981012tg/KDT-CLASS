```python
import cv2

image = cv2.imread('./data/like_lenna.png', cv2.IMREAD_GRAYSCALE)

if image is not None:
    print('read!')
else:
    print('not!')

print(type(image))
# 이미지를 읽었는데 웬 array? --> 텐서화 된것
```

```python
import matplotlib.pyplot as plt

plt.imshow(image, cmap="gray")
plt.title("Like Lenna")
plt.axis("off")
plt.show()
```

```python
print(image.shape)
```

```python
image_small = cv2.resize(image, (100, 100))

plt.imshow(image_small, cmap="gray")
plt.title("Small Like Lenna")
plt.axis("off")
plt.show()
```

```python
image_flip_tu = cv2.flip(image, 0)

plt.imshow(image_flip_tu, cmap="gray")
plt.title("Flipped Like Lenna")
plt.axis("off")
plt.show()
```

```python
image_flip_rl = cv2.flip(image, 1)

plt.imshow(image_flip_rl, cmap="gray")
plt.title("Flipped Like Lenna")
plt.axis("off")
plt.show()
```

```python
height, width = image.shape
matrix = cv2.getRotationMatrix2D((width/2, height/2), 90, 1)
res = cv2.warpAffine(image, matrix, (width, height)) # 순서 바꾼거 중요!

plt.imshow(res, cmap="gray")
plt.title("Rotated Like Lenna")
plt.axis("off")
plt.show()
```

```python
plt.imshow(image[20:100, 50:100], cmap="gray")
plt.title("Scrapped Like Lenna")
plt.axis("off")
plt.show()
```
