```python
# #함수의 구조
# def 함수이름():                def=define
#     수행할문장1
#     수행할문장2
#     ...
```

```python
def add(a,b):    #여기서 a,b를 매개변수(parameter)라고 함, 내 맘대로 이름 적어도 됨
    return a+b
```

```python
add(3,4)
```

```python
#1. 일반적인 함수
## 매개변수(parameter) O, 반환값(result) O

# def 함수이름(매개변수):
#     수행할문장
#     ...
#     return 반환값

def add(a,b):
    result=a+b
    return result

a= add(3,4)
print(a)
```

```python
#2. 입력값이 없는 함수
## 매개변수 X, 반환값 O

def say():
    return "hi"

a= say()
print(a)
```

```python
#3. 반환값이 없는 함수
## 매개변수 O, 반환값 X 

def add(a,b):
    print(f"{a},{b}의 합은 {a+b}")
    #return None --> return이 안적혀져 있으면, 이게 생략된거임

c=add(3,4)  #return값이 없으므로 c는 값이 없음


print(c)
```

```python
#4. 입력값도 없고 반환값도 없는 함수
## 매개변수 X, 반환값 X 

def say():
    print("hi")

a=say()
print(a)
```

```python
#매개변수를 지정하여 호출하기
def sub(a,b):
    result=a-b
    return result

c=sub(a=7,b=3)
print(c)
```

```python
#입력값이 몇개가 될지 모를때는 어떻게 해야할까?

# def 함수이름(*매개변수):
#     수행할문장...

def add_many(*args):      #args : arguments(인수), 굳이 args라고 안해도 되긴한데, 개발자들끼리 convention임
    result=0
    print(args)

    for i in args:
        result=result+i

    return result


result=add_many(1,2,3,4,5)
print(result)
```
