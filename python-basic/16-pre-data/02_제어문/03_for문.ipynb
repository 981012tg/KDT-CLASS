```python
marks = [90,25,67,45,80]

number=0  #기본값 세팅이 필요함

for mark in marks:
    number = number +1
    if mark >= 60:
        print(f"{number} 학생은 합격입니다.")
    else:
        print(f"{number} 학생은 불합격입니다,")
```

```python
#for문, continue
marks = [90,25,67,45,80]
number=0
for mark in marks:
    number +=1
    if mark <60:
        continue
    
    print(f"{number}학생은 합격입니다.")
```

```python
#for 문과 함께 자주 사용하는 range 함수

a=range(10)
print(a)
print(list(a))
```

```python
b=range(1,10) #이러면 10 포함 안함, 그 전까지
print(b)
print(list(b))
```

```python
c=range(1,10+1,2)
print(c)
print(list(c))
```

```python
d=range(10,1,-1) #거꾸로 하려면 반드시 -1,-2등 3번째 요소 필요
print(d)
print(list(d))
```

```python
#for와 range를 이용한 구구단
for i in range(2,9+1):
    for j in range(1,9+1):
        print(f"{i}X{j}={i*j}")
```

```python
#리스트 컴프리헨션 사용하기
a=[1,2,3,4]

result=[]

for num in a:
    result.append(num*3)

result
```

```python
#더 간단하게 하는법

a=[1,2,3,4]
result=[num*3 for num in a]
result
```

```python
#for문과 break문
for i in range(10): #range(0,10)
    if i==5:
        break
    print(i)
```

```python
#enumerate() 함수
fruits= ["apple", "banana", "orange"]
for fruit in fruits:
    print(f"과일: {fruit}")
```

```python
#enumerate() 함수를 씌우면 index도 같이 준다
fruits= ["apple", "banana", "orange"]
for fruit in enumerate(fruits):
    print(f"과일: {fruit}")
```

```python
#enumerate() 함수를 씌우면 index도 같이 준다
fruits= ["apple", "banana", "orange"]
for i, fruit in enumerate(fruits):
    print(f"{i+1}번째 과일: {fruit}")

```

```python
# zip함수로 여러 리스트 순회하기
names=["홍길동", "김철수", "이영희"]
scores=[85,92,78]

for name,score in zip(names,scores):
    print(f"{name}:{score}점")
```
