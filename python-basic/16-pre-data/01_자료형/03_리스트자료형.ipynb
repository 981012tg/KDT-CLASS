```python
#리스트는 어떻게 만들고 사용할까?

#리스트명=[요소1, 요소2,...]

a=[] # 비어있는 리스트
b=[1, 2, 3]
c=["life", "is",  "too", "short"]
d=[1, 2, "life", "is"]
e=[1, 2, ["life", "is"]]

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

```python
#리스트의 인덱싱과 슬라이싱
a=[1, 2, 3] #여기서 1을 가져 오려면?
print(a[0]) #문자열 처럼, 처음부터 인덱스가 0으로 시작
print(a[2])
print(type(a[2])) #값이 리스트에서 오긴 했지만, 값 자체는 숫자 이므로, int로 나옴
print(a[0] + a[2])
a[-1]
```

```python
b=['1', '2', '3', 'a', 'b', 'c'] #숫자로 보이지만, ''로 감쌌으므로, 문자로 취급됨
print(b[-1])
print(b[0])
print(b[2])
print(b[0]+b[2]) #문자로 취급되므로, 1+3=4가 아니라 그냥 결합이 되어 13이됨
```

```python
c=[1, 2, ['a', 'b', ['Life', 'is']]] #여기에 요소는? 3개
print(len(c))

#여기서 Life를 추출하려면?
print(c[-1])
print(c[-1][-1][0]) #chaining이라고 함
```

```python
#리스트의 슬라이싱
a=[1, 2, 3, 4, 5]
print(a[0:2]) #0은 써도 되고 안써도됨
print(a[3:3]) #둘이 같은 숫자면 아무것도 안가져옴
print(a[2:])
print(a[:3])
```

```python
a=[1, 2, 3, ['a', 'b', 'c'], 4, 5]

#['b', 'c']를 가져오려면?
len(a)
print(a[3][1:])
```

```python
#리스트 더하기
a=[1, 2, 3]
b=[4, 5, 6]
print(a+b)
```

```python
#리스트 곱하기 (a*b는 안됨)
a=[1, 2, 3]
print(a*3)
```

```python
#리스트의 수정
a=[1, 2, 3]
a[2] = 4
print(a)

```

```python
#del로 리스트 요소 삭제하기
a=[1, 2, 3]
del a[1]
print(a)
```

```python
a=[1,2,3,4,5]
del a[2:]
print(a)
```

```python
#리스트 관련 함수
##리스트 요소 추가하기 - append (많이 쓴다!)

a=[1,2,3] #여기다 4를 추가할거임
a.append(4)
print(a)
```

```python
a=[1,2,3,4]
a.append([5,6])
```

```python
#리스트 정렬 - sort(오름차순, 내림차순)
a.sort()
print(a)
a.sort(reverse=True)
print(a)
```

```python
a= [1,4,2,3]
sorted(a)
sorted(a,reverse=True)
```

```python
#리스트 뒤집기 - reverse
a=['a','c','b']

a.reverse()
print(a)
```

```python
#인덱스 반환 - index
a=['a','c','b']
print(a.index("c"))
```

```python
#리스트에 요소 삽입 - insert
a=[1,2,3]
a.insert(0,4) #0번째에 4를 집어넣겠다
print(a)

#제일 끝에 반복해서 삽입해야하는 경우에는 그렇게 자주 안쓰임
```

```python
#리스트에 요소 제거 - remove
a=[1,2,3,1,2,3]
a.remove(3) #첫번째 3만 지워짐
a.remove(3) #둘다 지우려면 한번 더 해야함

print(a)
```

```python
#리스트 요소 꺼내기 - pop
a=[1,2,3]
result=a.pop() #아무것도 안넣으면 마지막걸 꺼냄
print(result)
print(a) #위에서 마지막 값을 꺼냈으므로, 기존의 리스트에선 없음
```

```python
#리스트에 포함된 요소 x의 개수 세기 - count
a=[1,2,3,1]
a.count(1) #1의 개수
```

```python
#리스트 확장 - extend

a=[1,2,3]
a.extend([4,5])
print(a)

#append로 위에서 했을땐, 하나의 요소로 [4,5] 자체가 들어가버렸음. 더하기 연산과 비슷하다고 보면 됨
```
