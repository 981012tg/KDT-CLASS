```python
#딕셔너리는 어떻게 만들까?
#{key1: value1, key: value2, ...}

dic={
    'name' : 'pey',
    'phone': '010-1234-1234',
    'birth': '1118'
}

print(dic)
print(type(dic))
```

```python
# 딕셔너리 쌍 추가하기
a={1:"a"}
a[2]='b' #여기서 2는 인덱스 넘버가 아니라, key 값임.

print(a)
```

```python
a={1:"a"}
a[2]='b' #여기서 2는 인덱스 넘버가 아니라, key 값임.
a[3]=[1, 2, 3]
print(a)
```

```python
#딕셔너리 요소 삭제하기
a={1:"a"}
a[2]='b' #여기서 2는 인덱스 넘버가 아니라, key 값임.
a[3]=[1, 2, 3]
del a[1]
print(a)
```

```python
# 딕셔너리에 key를 사용해서 value 얻기
grade={
    'pey':10,
    'juli':99
}

print(grade['pey'])
print(grade['juli'])
```

```python
# 딕셔너리 관련 함수
## 1. key 리스트 만들기 - keys

a={
    'name' : 'pey',
    'phone': '010-1234-1234',
    'birth': '1118'
}

print(a.keys())
list(a.keys())
```

```python
## 2. value 리스트 만들기 - values

print(a.values())
list(a.values())
```

```python
## 3. key와 value 쌍 얻기 - items
print(a.items())
list(a.items()) #하나의 요소는 tuple로 되어있음
```

```python
# key, value의 쌍 모두 지우기
a.clear()
print(a)
```

```python
# key로 value 얻기 - get
a={
    'name' : 'pey',
    'phone': '010-1234-1234',
    'birth': '1118'
}
print(a['name'])
print(a.get('name'))
```

```python
#print(a['hobby']) 없는걸 가지고 오려면 에러가 난다
print(a.get('hobby')) #없는걸 가지고 오려면 None이라고 없다고 나온다
print(a.get('hobby',0)) #찾아서 없으면 0으로 대체해라는 말
```

```python
# 해당 key가 딕셔너리 안에 있는지 조사하기 - in
print(a)
print('name' in a)
print('email' in a)
```

```python
#key로 value값 꺼내기 - pop
a={
    'name' : 'pey',
    'phone': '010-1234-1234',
    'birth': '1118'
}
result=a.pop("phone")
print(result)
print(a)
```

```python
#리스트는 mutable 하므로 즉, 변경 가능하므로 딕셔너리의 키값으로 올 수 없음

a=[1,2,3]
a[0]=3
print(a)
```

```python
a=1
a=2
print(a)

#그럼 얘는 왜 가능하냐?
#이건 변경이 아니라 재할당이라고 봐야함
```
