```python
#튜플은 어떻게 만들까? tuple
t1=()
t2=(1,) #튜플의 가장 큰 특징은, 수정하거나 삭제할 수 없다는것, 그리고 하나의 요소를 적더라도 콤마(,)를 반드시 적어야한다는것
t3=(1,2,3) #2개 이상이면 마지막에 콤마 안해도 된다
t4=1,2,3 # 괄호 생략도 가능하다 --> 따라서, t2에서 컴마를 안하면 그냥 int로 나옴
t5=('a', 'b', ('ab', 'cd')) # 튜플 안에 튜플도 가능하다

print(type(t1))
print(type(t2))
print(type(t3))
print(type(t4))
print(type(t5))
print(t1)
print(t2)
print(t3)
print(t4)
print(t5)
```

```python
#튜플 인덱싱, 슬라이싱
t1=(1,2,'a','b')
print(t1[2])
print(t1[-1])
print(t1[0:2])
print(t1[1:])
```

```python
#튜플 더하기
t1=(1,2,'a','b')
t2=(3,4)
print(t1)
print(t2)
print(t1+t2)
#t1, t2 자체가 변한게 아니므로 이건 가능함
```

```python
#튜플 곱하기
print(t2*3)
print(t2)
#역시 t2 자체가 변한게 아니므로 가능함
```
