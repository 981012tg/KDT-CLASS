```python
# 그룹 객체 만들기 (분할단계)
# DataFrame객체.groupby(기준이 되는 열)

import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age', 'sex', 'class', 'fare', 'survived']]

df.head()
```

```python
grouped = df.groupby(['class'], observed=True)
print(grouped)
```

```python
for key, group in grouped:
    print('* key :', key)
    print('* number :', len(group))
    print(group.head())
    print('\n')
```

```python
average = grouped.mean(numeric_only=True)
average
```

```python
group3 = grouped.get_group(('Third',))
group3.head()
```

```python
# 여러 열을 기준으로 그룹화
# DataFrame객체.groupby(기준이 되는 열의 리스트)

grouped_two = df.groupby(['class', 'sex'], observed=True)

for key, group in grouped:
    print('* key :', key)
    print('* number :', len(group))
    print(group.head())
    print('\n')
```

```python
average_two = grouped_two.mean(numeric_only=True)
average_two
```

```python
# 모든 열에 여러 함수를 매핑 : group 객체.agg([함수1, 함수2, 함수3, ...])
# 각 열마다 다른 함수를 매핑 : group 객체.agg({'열1' : 함수1, '열2' : 함수2, ...})

agg_all = grouped.agg(['min', 'max'])
agg_all
```

```python
agg_sep = grouped.agg({'fare' : ['min', 'max'], 'age' : 'mean'})
agg_sep
```

```python
def min_max(x):
    return x.max() - x.min()

agg_minmax = grouped[['age', 'fare']].agg(min_max)

agg_minmax
```
