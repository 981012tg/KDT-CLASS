```python
#문자열 메서드
import pandas as pd
import numpy as np

ser = pd.Series([
    "Apple_사과", "Banana_바나나", "Cherry_체리", np.nan],
    index = ["First ", " Second", " Third ", "Fourth"])

print(ser)
print("\n")

ser2 = pd.Series([
    "Apple_사과",
    "Banana_바나나",
    "Cherry_체리",
    np.nan
    ],
    index = [
        "First ",
        "Second ",
        "Third ",
        "Fourth "
	], dtype = "string"
)

print(ser2)
```

```python
print(ser.str.lower())
print('\n')

print(ser.str.upper())
print("\n")

print(ser.str.len())
```

```python
print(ser.str.split("_"))
print("")
print(ser.str.split("_", expand=True))
print("")
print(ser.str.split("_").str.get(1))
```

```python
idx = ser.index
print(idx)

print(idx.str.strip())
print(idx.str.lstrip())
print(idx.str.rstrip())
```

```python
print(ser.str.replace("_", ":", regex=False))
print("")

print(ser.str.replace("[^a-zA-Z\s]", "", regex=True))
```

```python
print(ser.str[0])
print("\n")

print(ser.str[0:4])
```

```python
contains_A = ser.str.contains("A", na=False)
print(contains_A)
print("")

contains_patern = ser.str.contains(r"[A|B][a-z]+")
print(contains_patern)
print("")

fullmatch_pattern = ser.str.fullmatch(r"[A|B][a-z]+")
print(fullmatch_pattern)
```

```python
ser3 = pd.Series([
    "a1",
    "b2",
    "c3",
    "d4a5",
    "e6e7",
    "f8"], 
dtype="string")

digits = ser3.str.extractall(r"(\d)").groupby(level=0).agg(''.join)
print(digits)
print("")

letters = ser3.str.extractall(r"([a-z])").groupby(level=0).agg(''.join)
print(letters)
print("")

split_columns = ser3.str.extract(r"(?P<Letter>[a-z]+)(?P<Digit>\d+)")
print(split_columns)
```
