```python
import numpy as np #numpy ->np, pandas ->pd 이거 convention임. 딴걸로 해도 되긴 한데, 따르는게 좋다

arr1 = np.array([1, 2, 3, 4, 5])

print(arr1)
print(type(arr1))
```

```python
arr2 = np.array([
    [1,2,3],
    [4,5,6]
])

print(arr2)
print(type(arr2))
```
