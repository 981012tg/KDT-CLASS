```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import KFold
#KFold : 데이터를 여러 조각으로 나누어서 여러번 학습하고 여러번 평가하는 교차검증 방법

from sklearn.datasets import load_iris
import numpy as np

iris = load_iris()
features = iris.data #X값 - 꽃 특징
label = iris.target #y값 - 0,1,2
dt_clf = DecisionTreeClassifier(random_state = 42)

kfold = KFold(n_splits = 5)
# 전체 데이터를 K개의 덩어리 즉, Fold로 나눈다.
# 여기서는 5개의 fold로 나눈다.
# 30 + 30 + 30 + 30 + 30 --> 이중 30개는 test용, 나머지는 train용
# 근데 이걸 5번 돌아가면서 test용으로 사용
# 예를들어 fold1을 test용으로 두고 fold2~5를 train
# fold2를 test용으로 두고 또 나머지를 train --> 이런식으로 계속 학습
cv_accuracy = []
#교차 검증을 수행하면서 나온 정확도들을 저장할 리스트를 하나 만든다
print(features.shape[0])
```

```python
n_iter = 0
# 현재 몇번째 K-Fold cv를 하고 있는지 카운트

for train_index, test_index in kfold.split(features):
	# kfold.split(features)
	# --> features 데이터를 기준을 5번의 train/test 인덱스 조합을 만든다
	# 데이터가 150개이므로 0~149까지 인덱스가 있음
	X_train, X_test = features[train_index], features[test_index]
	# 방금 받은 train_index와 test_index를 이용해서 실제 x 데이터를 가져온다
	y_train, y_test = label[train_index], label[test_index]
	# y도 똑같은 index로 가져온다.

	dt_clf.fit(X_train, y_train)
	#학습시킨다
	pred = dt_clf.predict(X_test)
	#테스트 문제를 줘서 풀게한다.
	n_iter += 1
	# 5번까지 반복한다.

	accuracy = np.round(accuracy_score(y_test, pred), 4)
	# 정확도를 측정해서, 소수 넷째자리 까지 나오게 반올림한다
	train_size = X_train.shape[0]
	# X_train의 행 개수 확인 --> train은 120개
	test_size = X_test.shape[0]
	# test 는 30개
	print(n_iter, accuracy, train_size, test_size)
	# 현재 회차의 결과를 출력한다.예를들어, 1회차, 정확도 1.0, 학습데이터 120개, 테스트데이터 30개
	cv_accuracy.append(accuracy)
	# 그리고 정확도를 처음에 만들어놨던 빈 리스트에다가 저장, 회차마다 추가함

print(np.mean(cv_accuracy))
# for문이 다 돌면, 빈 리스트에서, 5개의 요소를 가진 리스트가 됨. 그 리스트의 평균을 낸 것
# --> '대략 이정도의 성능을 낸다'는것을 알 수 있음
```
