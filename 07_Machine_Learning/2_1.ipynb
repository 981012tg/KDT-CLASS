```python
# 필요한 라이브러리 불러오기
# scikit-learn은 Python에서 머신러닝 작업을 쉽게 수행하도록 만들어진 라이브러리

import sklearn 

from sklearn.datasets import load_iris
# sklearn 안의 datasets에서 load_iris를 가져온다
# datasets는 연습이나 테스트에 활용할 수 있는 데이터셋 관련 기능이 있음
# load_iris는 그 중 Iris(붓꽃) 데이터셋을 불러오는 함수
 #데이터를 가져오는 함수


from sklearn.tree import DecisionTreeClassifier
# sklearn 안의 tree 에서, DecisionTreeClassfier를 가져온다
# tree는 Decision Tree 관련 기능이 들어있음
# DecisionTreeClassifier는 의사결정나무를 이용해서 분류하는 머신러닝모델이다.
# 예를들어, 붓꽃 데이터(ex)꽃받침 길이, 너비, 꽃잎 길이, 너비 등등)을 보고
# 얘가 setosa, versicolor, virginica중 뭔지 판단한다
# 머신러닝 모델 객체를 만드는 클래스

from sklearn.model_selection import train_test_split
# sklearn 안의 model_selection에서, train_test_split를 가져와라
# modle_selection은, 모델을 학스하고 평가할 데이터를 나누거나 검증하는데 필요한 기능들이 들어있다.
# train_test_split()은 데이터를 학습용과 테스트용으로 나누는 함수
```

```python
import pandas as pd

iris = load_iris()
# iris에 iris 데이터 전부를 가져와서 저장

## Iris 데이터 셋에서 피처(X 값)
iris_data = iris.data
# Iris 데이터 중 머신러닝 모델에게 입력할 데이터를 가져온다
# Iris의 데이터에는 꽃받침길이, 꽃받침너비, 꽃잎길이, 꽃잎너비가 들어있다
# [5.1, 3.5, 1.4, 0.2] <-- 이런식으로 되어있고, 순서대로 위의 값이다.
# 총 150개의 데이터가 있다. --> 150행, 4열 형태

# Iris 데이터 셋에서 Y값
iris_label = iris.target
# 정답 데이터를 가져온다.
# 예를들어, X가 [5.1, 3.5, 1.4, 0.2], y가 0이라면,
# 꽃의 특징이 X [5.1, 3.5, 1.4, 0.2]인 꽃은 0번 꽃이라는 소리

print(iris_label) #0, 1, 2 만 존재 --> 꽃 종류가 3개다
print(iris.target_names) # 그 꽃의 이름들

iris_df = pd.DataFrame(
    data = iris_data,
    columns = iris.feature_names
    )
iris_df['label'] = iris.target
# label이라는 새로운 column을 만들고, 거기에는 iris.target
# 즉, 얘가 몇번 꽃인지를 나타낸다. --> 정답지를 제공하는것
iris_df.tail()
```

```python
#Iris 데이터 150개를 ML 학습용과 성능 평가용으로 나누는 부분
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label,
                                                    test_size = 0.2, random_state = 11)

# train_test_split()은, 데이터를 학습용과 테스트용으로 나누는 함수
# 모델에게 모든 데이터를 학습시킨 다음, 그 데이터를 다시 시험문제로 내면, 모델이 진짜 그걸 알고 푸는건지,
# 아니면 외워서 푼건지 확인할 수 없음 --> 그래서 처음부터 나눈다.
# test_size = 0.2 라는 말은, 전체 데이터 중 20프로를 테스트 데이터로 사용하겠다는뜻
# 즉, 150개중 120개로 학습하고 30개로 테스트를 한다
# 이 함수는 기본적으로 데이터를 랜덤하게 섞어서 train과 test를 나누는데,
# 아무 설정도 하지 않으면, 실행할때마다 데이터가 다르게 나뉜다. 
# 예를들어, 첫번째 실행에서 1번 꽃은 train용이었는데, 두번째 실행에서는 test용으로 될 수도 있음
# 이러면 정확도가 조금씩 달라지게됨
# random_state = 11이라고 하면, 랜덤하게 나누긴 하지만 그 랜덤 결과를 재현할 수 있도록 고정
```

```python
dt_clf = DecisionTreeClassifier(random_state = 11)
# 의사결정나무 분류모델을 만들고 그 모델을 변수에 저장함
# 모델 학습 결과를 재현할 수 있도록 random_state는 고정

dt_clf.fit(X_train, y_train);
# 실제로 학습시키는 부분
# fit에 X_train, y_train 즉, 학습용 데이터를 준다
# 이를 통해 규칙을 학습한다
```

```python
pred = dt_clf.predict(X_test)
## print(pred)

# 학습이 끝난 모델에게 테스트 문제를 풀게한다
# dt_clf는 이제, 이전 셀에서 학습용 데이터로 공부한 의사결정나무 모델
# 이젠 X_test를 주고 이전에 공부했던 내용을 바탕으로 어떤 꽃인지 예측하게 한다
# 이때는 y_test는 주지 않는다. y_test는 답지이기 때문
```

```python
from sklearn.metrics import accuracy_score
# sklearn의 metrics 에서 accuracy_score 함수를 가져온다
# metrics는 머신러닝 모델의 성능을 평가하는 여러 방법이 들어있는 모듈
# accuracy_score()는 그 중 accuracy(정확도)를 계산하는 함수

print(accuracy_score(y_test, pred))
# y_test(실제정답)과 pred(모델이 예측한 정답)을 비교한다
# 맞춘 비율을 나타낸다.(여기서는 30개중 28개 정답 -> 93.3%)
```
