```python
import pandas as pd
```

```python
#딕셔너리를 시리즈로 변환
dict_data = {
    "a" : 1,
    "b" : 2,
    "c" : 3
}

sr = pd.Series(dict_data)

print(type(sr))

print(sr)
```

```python
#시리즈의 인덱스

list_data = ['2019-01-02', 3.14, 'ABC', 100, True]

sr = pd.Series(list_data)


print(type(sr))
print(sr)
```

```python
print(sr.index)
print(sr.values)
print(sr.dtype)
```

```python
#배열의 크기
print(len(sr))

# shape
print(sr.shape)

# ndim
print(sr.ndim)
```

```python
import pandas as pd

tup_data = ('영인', '2010-05-01', '여', True)
sr = pd.Series(tup_data, index=['이름', '생년월일', '성별', '학생여부'])
print(sr)
```

```python
# 시리즈 원소 선택
print(sr["이름"]) #인덱스로하면 지금 버전으로 하면 오류남(버전이 너무 높음)
```

```python
sr[['생년월일', '성별']] #인덱스로하면 지금 버전으로 하면 오류남(버전이 너무 높음)
```
