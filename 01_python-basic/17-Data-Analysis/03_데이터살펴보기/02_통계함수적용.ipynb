```python
# 평균값

import pandas as pd

df = pd.read_csv('./data/auto-mpg.csv', header=None)

df.columns = [
    'mpg',
    'cylinders',
    'displacement',
    'horsepower',
    'weight',
    'acceleration',
    'model year',
    'origin',
    'name'
]

print(df.mean(numeric_only=True))
print("")
print(df['mpg'].mean())
print(df.mpg.mean())
print("")
print(df[['mpg', 'weight']].mean)
```

```python
# 중앙값
print(df.median(numeric_only=True))
print("")
print(df['mpg'].median())
```

```python
#최댓값
print(df.max(numeric_only=True))
print("")
print(df['mpg'].max())
```

```python
#최솟값
print(df.min(numeric_only=True))
print("")
print(df['mpg'].min())
```

```python
#표준편차
print(df.std(numeric_only=True))
print("")
print(df['mpg'].std())
```

```python
#상관계수
print(df.corr(numeric_only=True))
print("")
print(df[['mpg', 'weight']].corr())
```
