## 문제 ① 결측치 처리 및 평균 비교

`df` 데이터에 대해서 **결측치가 가장 많은 변수**를 찾아서 해당 결측치를 **0으로 대치**하고,  
**결측치를 제외한 평균**과 **0으로 대치한 후의 평균**과의 차이를 구하시오.

> 출력: 두 평균의 차이 (절대값 여부는 풀이 방식에 따름)

````python
print(df.isnull().sum())
before_ozone = df["Ozone"].mean()
before_ozone

df["Ozone"] = df["Ozone"].fillna(0)
after_ozone = df["Ozone"].mean()

result_1 = before_ozone - after_ozone
result_1 = round(result_1, 3)
result_1
````

## 문제 ② Min-Max / Z 정규화 평균 비교

`Wind` 변수에 대해서 **Min-Max 정규화**를 수행한 후 평균값과  
**Z 정규화(표준화)**를 수행한 후 평균값의 차이를 구하시오.

- Min-Max 정규화: $(x - min) / (max - min)$
- Z 정규화: $(x - mean) / std$

> 출력: 두 평균의 차이

````python
df["Wind"].head()

wind_min=df["Wind"].min()
wind_max=df["Wind"].max()
wind_mean=df["Wind"].mean()
wind_std=df["Wind"].std()

df["meanmax"] = (df["Wind"] - wind_min) / (wind_max - wind_min)

df.head()

df["zscore"] = (df["Wind"] - wind_mean) / (wind_std)

df.head()

result2 = df["meanmax"].mean() - df["zscore"].mean()
result2 = round(result2, 2)

result2
````
