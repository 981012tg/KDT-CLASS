```python
# 데이터 프레임 연결
# pandas.concat(데이터프레임의 리스트)

import pandas as pd
```

```python
dict1 = {
    "a" : ["a0", "a1", "a2", "a3"],
    "b" : ["b0", "b1", "b2", "b3"],
    "c" : ["c0", "c1", "c2", "c3"]}

df1 = pd.DataFrame(dict1, index = [0,1,2,3])
df1
```

```python
dict2 = {
    "a" : ["a2", "a3", "a4", "a5"],
    "b" : ["b2", "b3", "b4", "b5"],
    "c" : ["c2", "c3", "c4", "c5"],
    "d" : ["d2", "d3", "d4", "d5"],}

df2 = pd.DataFrame(dict2, index=[2, 3, 4, 5])
df2
```

```python
result1 = pd.concat([df1, df2])
result1
```

```python
result2 = pd.concat([df1, df2], ignore_index=True)
result2
```

```python
result2_in = pd.concat([df1, df2], axis=0, join='inner')
result2_in
```

```python
result3 = pd.concat([df1, df2], axis=1)
result3, '\n'
```

```python
result3_in = pd.concat([df1, df2], axis=1, join='inner')
result3_in
```

```python
sr1 = pd.Series(['e0', 'e1', 'e2', 'e3'], name='e')
sr2 = pd.Series(['f0', 'f1', 'f2'], name='f', index=[3, 4, 5])
sr3 = pd.Series(['g0', 'g1', 'g2', 'g3'], name='g')
sr4 = pd.Series(['a6', 'b6', 'c6', 'd6'], index=['a', 'b', 'c', 'd'])

result4 = pd.concat([df1, sr1], axis=1)
result4
```

```python
result5 = pd.concat([df2, sr2], axis=1)
result5
```

```python
result6 = pd.concat([df2, sr4.to_frame() .T], ignore_index=True)
result6
```

```python
result7 = pd.concat([sr1, sr3], axis=1)
result7
```

```python
result8 = pd.concat([sr1, sr3], axis=0)
result8
```

```python
df1 = pd.read_excel('./data/stock_price.xlsx')
df1
```

```python
merge_outer = pd.merge(df1, df2, how='outer', on='id')
merge_outer
```

```python
merge_left = pd.merge(df1, df2, how='left',
                      left_on=['id', 'stock_name'], right_on=['id', 'name'])
merge_left
```

```python
merge_right = pd.merge(df1, df2, how='right',
                      left_on=['id', 'stock_name'], right_on=['id', 'name'])
merge_right
```

```python
merge_cross = pd.merge(df1, df2, how='cross')
merge_cross
```

```python
sdf1 = pd.DataFrame({
    'employee': ['Alice', 'Sam', 'Eva'],
    'department': ['HR', 'Tech', 'HR']
    })
sdf2 = pd.DataFrame({
    'employee': ['Eva', 'Alice', 'Sam'],
    'start_year': ['2018', '2019', '2020']
    })

result_one_to_one = pd.merge(sdf1, sdf2, on='employee')
result_one_to_one
```

```python
sdf3 = pd.DataFrame({
    'department': ['HR', 'Tech'],
    'manager' : ['Tina', 'Alex']
})

result_many_to_one = pd.merge(sdf1, sdf3, on='department')
result_many_to_one
```

```python
sdf4 = pd.DataFrame({
    'department': ['HR', 'HR', 'Tech', 'Tech', 'Finance'],
    'task' : ['recruiting', 'payroll', 'development', 'support', 'budgeting']
})

result_many_to_many = pd.merge(sdf1, sdf4, on='department')
result_many_to_many
```

```python
import requests
import pandas as pd

my_key = "dpd-18545c69-0bbe8c1e0c8abc6954ae9fba80d70c40b363199406e6d75dd2f288a7a66f3cf1"

def get_financials(stockCode, indicatorName, apiKey, consolidated=True, ttm=True):
    
	if consolidated:
		if ttm:
			url = f"https://api.dapada.io/company/getConsolidatedFinancialDataByTTM?indicatorName={indicatorName}&stockCode={stockCode}"
		else:
			url = f"https://api.dapada.io/company/getConsolidatedFinancialDataByCUR?indicatorName={indicatorName}&stockCode={stockCode}"
	else:
		if ttm:
			url = f"https://api.dapada.io/company/getSeparatedFinancialDataByTTM?indicatorName={indicatorName}&stockCode={stockCode}"
		else:
			url = f"https://api.dapada.io/company/getSeparatedFinancialDataByCUR?indicatorName={indicatorName}&stockCode={stockCode}"

	headers = {"Authorization" : f"{apiKey}"}
	response = requests.get(url, headers = headers)
	result = response.json()
	return pd.DataFrame(result)

df_revenue = get_financials(stockCode = '005930',
							indicatorName='매출액',
							apiKey = my_key,
							consolidated=True,
							ttm=False)

df_revenue.columns = ['매출액', '분기']

df_revenue = df_revenue.set_index('분기', drop=True)

df_revenue = df_revenue[df_revenue.index.str.startswith('2022')]

df_revenue
```

```python
df_margin = get_financials(stockCode = '005930',
							indicatorName='영업이익',
							apiKey = my_key,
							consolidated=True,
							ttm=False)

df_margin.columns = ['영업이익', '분기']

df_margin = df_margin.set_index('분기', drop=True)

df_margin = df_margin[df_margin.index.str.endswith('Q3')]

df_margin
```

```python
df_revenue_margin=df_revenue.join(df_margin)
df_revenue_margin
```
