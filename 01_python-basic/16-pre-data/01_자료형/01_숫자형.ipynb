```python
#숫자형은 어떻게 만들고 사용할까?
a=1234 #여기서 a,b,c는 변수
b=-1234 #등호의 의미는, 같다의 의미 보다는, 대입한다의 의미
c=0.1
print(a) 
print(b) 
print(c)
print(a+b)

print(type(a)) #type은 자료형을 알려주는 함수
#int는 정수형을 의미하는 자료형
print(type(b)) 
print(type(c))
```

```python
#이제는 실수형
a=1.2
b=-3.14

print(a)
print(b)

print(type(a))
print(type(b))
```

```python
#8진수, 16진수, 알아만 두자
a=0o177 #8진수
print(a)
print(1*(8**2)+7*(8**1)+7*(8**0))
b=0x1A #16진수
print(b)
```

```python
a=0x8ff #16진수 8ff는 8*16^2 + 15*16^1 + 15*16^0
b=0x1A #16진수 1A는 1*16^1 + 10*16^0
print(a)
# a=10
# b=11    
# c=12
# d=13
# e=14
# f=15
```

```python
#사칙연산
a=3
b=4

print(a+b) #덧셈
print(a-b) #뺄셈
print(a*b) #곱셈
print(a/b) #나눗셈
```

```python
#제곱수
a=3
b=4

print(a**b) #3의 4제곱
```

```python
#나눗셈 후 나머지 반환 %, 몫 반환 //
print(7%3) #7을 3으로 나눈 나머지는 1
print(7/3) #7을 3으로 나눈 결과는 2.333...
print(7//3) #7을 3으로 나눈 몫은 2
```

```python
#복합연산자

a=3
#a=a+1 #a에 1를 더해서 a에 다시 대입한다
a+=1 #a에 1를 더해서 a에 다시 대입한다 --> 둘은 같은의미
print(a)
```
