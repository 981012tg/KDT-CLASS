```python
#문자열은 어떻게 만들고 사용할까?

## 1.큰따옴표로 양쪽 둘러싸기
a = "Hello World"

## 2.작은따옴표로 양쪽 둘러싸기
b = 'Python is fun'

## 3.큰따옴표 세 개로 양쪽 둘러싸기
c = """Life is too short, you need python"""

## 4.작은따옴표 세 개로 양쪽 둘러싸기
d = '''Life is too short, you need python'''

print(a)
print(b)
print(c)
print(d)

print(type (a))
print(type (b))
print(type (c))
print(type (d))
```

```python
#문자열 안에 작은 따옴표를 포함하기
a="Python's favorite food is perl" #문자열 안에 작은 따옴표가 포함되어 있어서 오류가 발생
b="""Python's favorite food is perl""" #큰 따옴표로 문자열을 감싸면 작은 따옴표를 포함할 수 있다. 
c='Python\'s favorite food is perl' #문자열 안에 작은 따옴표를 포함하기 위해서는 작은 따옴표 앞에 역슬래시(\)를 붙여야 한다.
print(a)
print(b)
print(c)
```

```python
#문자열 안에 큰 따옴표를 포함하기
a='"Python is very easy" he said.'
b=""""Python is very easy" he said. """ #큰 따옴표로 문자열을 감싸면 큰 따옴표를 포함할 수 있다.
c="\"Python is very easy\" he said." #문자열 안에 큰 따옴표를 포함하기 위해서는 큰 따옴표 앞에 역슬래시(\)를 붙여야 한다.
print(a)
print(b)
print(c)
```

```python
# escape code
# \n : 문자열 안에서 줄을 바꿀때
# \t : 문자열 안에서 탭 간격을 줄 때
# \\ : 문자열 안에서 \를 표현할 때
# \' : 문자열 안에서 작은 따옴표(')를 표현할 때
# \" : 문자열 안에서 큰 따옴표(")를 표현할 때
```

```python
# 문자열 연산
a="Python "
b="is fun"
print(a+b) #문자열 연결하기
print(a*3) #문자열 반복하기
```

```python
#문자열 길이 구하기 (엄청 많이 쓴다)
a = "Life is too short"
print(len(a))
len(a)
```

```python
#문자열 인덱싱과 슬라이싱
a = "Life is too short, you need python."
indexing = a[3] #인덱싱
slicing = a[12:17]#슬라이싱 (0부터 4까지, 4는 포함하지 않는다.)
print(indexing)
print(slicing)  
print(a[3])
print(len(a))
print(a[-3]) #뒤에서부터 인덱싱하기
print(a[33])
print(a[:]) #처음부터 끝까지 슬라이싱하기
print(a[:17]) #처음부터 16까지 슬라이싱하기, 0은 해도 되고 안 해도 된다.
print(a[19:]) #19부터 끝까지 슬라이싱하기, 끝은 해도 되고 안 해도 된다.
print(a[19:-1]) #19부터 뒤에서 1까지 슬라이싱하기, 뒤에서 1은 해도 되고 안 해도 된다.


```

```python
#문자열 포매팅 (예전에 많이 씀)
## 1. 숫자 바로 대입
"I eat %d apples." % 3   
print("I eat %d apples." % 3)

##2. 문자열 바로 대입
"I eat %s apples." % "five" 
print("I eat %s apples." % "five")
```

```python
##3. 숫자 값을 나타내는 변수로 대입
number = 3
print("I eat %d apples." % number)
```

```python
##4. 2개 이상의 값 넣기
number = 10
day = "three"
print("I ate %d apples. so I was sick for %s days." % (number, day))
```

```python
# format 함수를 사용한 포매팅
print( "I eat {0} apples.".format(3))
print( "I eat {0} apples.".format("five"))
```

```python
# format 함수를 이용해서 2개 이상의 값 넣기
number=10
day="three"
print("I ate {0} apples. so I was sick for {1} days." .format(number, day))
```

```python
# f 문자열 포매팅 (이걸 제일 많이씀. 아마 제일 편할듯)

name="홍길동"
age=10
print("나의 이름은 {name}입니다. 나이는 {age}입니다.")
print(f"나의 이름은 {name}입니다. 나이는 {age}입니다.")  #f를 앞에 붙여줘야함
```

```python
#문자열에 관련된 함수들
a="hobby"
print(type(a))

#문자 개수 세기 - count
print(a.count("b")) 
```

```python
#위치 알려주기 - find

a="Python is the best choice"
print(a.find("b")) #b가 처음 나온 위치를 알려줌, 처음이 0부터임
print(a.find("k")) #찾는 문자가 없으면 -1을 반환
```

```python
#위치 알려주기2 - index
a="Life is too short"
print(a.index("t")) #t가 처음 나온 위치를 알려줌, 처음이 0부터임
#print(a.index("k")) #찾는 문자가 없으면 에러가 남
```

```python
#문자열 삽입 - join
a=",".join("abcd") #abcd라는 문자열의 각각의 문자 사이에 ,를 삽입
print(a)
```

```python
# 소문자를 대문자로 바꾸기 - upper
a="hi"
a.upper()
```

```python
#대문자를 소문자로 바꾸기 - lower
a="HI"
a.lower()
```

```python
# 공백지우기
a="   hi   "
print(a)
print(a.strip()) #앞뒤 공백 제거
print(a.lstrip()) #왼쪽 공백 제거
print(a.rstrip()) #오른쪽 공백 제거
```

```python
#문자열 바꾸기 - replace
a="Life is too short"
print(a.replace("Life", "Your leg"))
```

```python
#문자열 나누기 - split
a="Life is too short"
print(a.split()) #공백을 기준으로 문자열 나누기
print(a.split("i")) #i를 기준으로 문자열 나누기
```
